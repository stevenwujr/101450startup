# Startup Profit Prediction with Multiple Linear Regression

This notebook follows the CRISP-DM workflow for the Kaggle 50 Startups dataset.

- Business Understanding: predict startup profit
- Data Understanding: inspect dataset and columns
- Data Preparation: one-hot encode State and split train/test
- Modeling: Linear Regression
- Evaluation: R², MAE, MSE, RMSE
- Deployment: save the model


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
url = 'https://raw.githubusercontent.com/harimittapalli/Mulitple-Linear-Reggression/master/50_Startups.csv'
data = pd.read_csv(url)
print(data.head())
print(data.columns)
print(data.shape)
```

```python
X = data.drop(columns=['Profit'])
y = data['Profit']

preprocessor = ColumnTransformer(
    transformers=[('cat', OneHotEncoder(drop='first'), ['State'])],
    remainder='passthrough'
)

X_processed = preprocessor.fit_transform(X)
print(X_processed[:2])

X_train, X_test, y_train, y_test = train_test_split(
    X_processed, y, test_size=0.2, random_state=42
)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print(f'R-squared: {r2:.4f}')
print(f'MAE: {mae:.4f}')
print(f'MSE: {mse:.4f}')
print(f'RMSE: {rmse:.4f}')
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(y_test, y_pred, alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', linestyle='--')
plt.title('Actual vs Predicted Profit')
plt.xlabel('Actual Profit')
plt.ylabel('Predicted Profit')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# CRISP-DM summary
print('Business understanding: predict startup profit')
print('Data preparation: one-hot encoding for State and train/test split')
print('Modeling: multiple linear regression')
print('Evaluation: R², MAE, MSE, RMSE')
print('Deployment: save the model for future use')
```
